```python
#Find the common customers between customers and new_customers
customers = {"Rahul", "Anita", "Kiran", "Sneha", "Vikram"}
new_customers = {"Rahul","pavi","hari"}
print(customers.intersection(new_customers))
```

```python
#Find customers that are present in either customers or new_customers but not both.
customers = { "Kiran", "Sneha", "Vikram"}
new_customers = {"Amit",  "Rahul", "Divya"}
print(customers.symmetric_difference(new_customers))
```
