### Q1. Create a account class for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

1. List the objects, their responsibilities, and their invariants before coding.
2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.
3. Keep balances private and expose controlled deposit/withdraw/transfer behavior.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
class Account:
    def __init__(self, account_number: str, balance: float = 0.0):
        if not account_number:
            raise ValueError("Account number cannot be empty.")
        if balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.account_number = account_number
        self._balance = float(balance)

    def deposit(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount
        return self._balance

    def withdraw(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError("Insufficient funds.")
        self._balance -= amount
        return self._balance

    def get_balance(self) -> float:
        return self._balance


# Demonstration
acc = Account("ACC-1001", 100.0)
acc.deposit(50.0)
acc.withdraw(30.0)
print(f"Account {acc.account_number} final balance: ${acc.get_balance():.2f}")
```

```python
from datetime import datetime

class Case2Transfer:
    def __init__(self, transfer_id: str, source: Case2Account, target: Case2Account, amount: float):
        self.transfer_id = transfer_id
        self.source = source
        self.target = target
        self.amount = amount
        self.timestamp = None
        self.status = "PENDING"

    def process(self) -> bool:
        if self.source.withdraw(self.amount):
            self.target.deposit(self.amount)
            self.status = "COMPLETED"
            self.timestamp = datetime.now()
            return True
        self.status = "FAILED"
        return False


# Demonstration
acc1 = Case2Account("ACC-1018A", 1000.0)
acc2 = Case2Account("ACC-1018B", 200.0)
t = Case2Transfer("TXN-1018", acc1, acc2, 300.0)
print(f"Success: {t.process()}, Status: {t.status}")
```

```python
class ProtectedAmountTransaction:
    def __init__(self, reference: str, amount: float):
        self.reference = reference
        self.amount = amount  # Triggers property setter

    @property
    def amount(self) -> float:
        return self._amount

    @amount.setter
    def amount(self, value: float) -> None:
        if not isinstance(value, (int, float)):
            raise TypeError("Amount must be a numeric value.")
        if value <= 0:
            raise ValueError("Transaction amount must be strictly greater than zero.")
        self._amount = float(value)


# Demonstration
txn = ProtectedAmountTransaction("ACC-1023", 250.75)
print(f"Transaction Reference: {txn.reference}, Validated Amount: ${txn.amount:.2f}")
```

```python
from datetime import date


class Account:
    def __init__(self, account_ref: str, balance: float = 0.0):
        if not account_ref:
            raise ValueError("Account reference cannot be empty.")
        if balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.account_ref = account_ref
        self._balance = float(balance)

    def withdraw(self, amount: float) -> bool:
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError("Insufficient funds.")
        self._balance -= amount
        return True

    def get_balance(self) -> float:
        return self._balance


class PerishableAccount(Account):
    def __init__(self, account_ref: str, balance: float, expiry_date: date):
        super().__init__(account_ref, balance)
        if not isinstance(expiry_date, date):
            raise TypeError("expiry_date must be a valid datetime.date object.")
        self.expiry_date = expiry_date

    def is_expired(self) -> bool:
        return date.today() > self.expiry_date

    def withdraw(self, amount: float) -> bool:
        if self.is_expired():
            raise PermissionError("Account has expired. Withdrawals are restricted.")
        return super().withdraw(amount)


# Demonstration
valid_account = PerishableAccount("ACC-1031A", 500.0, date(2030, 12, 31))
print(f"Is Expired: {valid_account.is_expired()}")
print(f"Withdrawal Success: {valid_account.withdraw(100.0)}")
print(f"Remaining Balance: ${valid_account.get_balance():.2f}")

expired_account = PerishableAccount("ACC-1031B", 300.0, date(2020, 1, 1))
print(f"Is Expired: {expired_account.is_expired()}")
try:
    expired_account.withdraw(50.0)
except PermissionError as e:
    print(f"Caught expected error: {e}")
```

```python
class Account:

    def __init__(self, account_ref: str, balance: float):
        if not account_ref:
            raise ValueError("account_ref required.")
        if balance < 0:
            raise ValueError("Initial balance cannot be negative.")

        self.account_ref = account_ref
        self._balance = float(balance)
        self._reserved = 0.0

    @property
    def available_balance(self) -> float:
        return self._balance - self._reserved

    @property
    def total_balance(self) -> float:
        return self._balance

    def reserve(self, amount: float) -> bool:
        if amount <= 0:
            raise ValueError("Reservation amount must be positive.")
        if amount > self.available_balance:
            raise ValueError("Insufficient available funds for reservation.")
        self._reserved += amount
        return True

    def release(self, amount: float) -> bool:
        if amount <= 0 or amount > self._reserved:
            raise ValueError("Invalid release amount.")
        self._reserved -= amount
        return True

    def commit_reservation(self, amount: float) -> bool:
        if amount <= 0 or amount > self._reserved:
            raise ValueError("Invalid commit amount.")
        self._reserved -= amount
        self._balance -= amount
        return True


# Demonstration
try:
    acc = Account("ACC-1056", 500.0)
    acc.reserve(150.0)
    print(
        f"Total: ${acc.total_balance:.2f}, Available: ${acc.available_balance:.2f}"
    )

    acc.commit_reservation(150.0)
    print(
        f"Post-commit Total: ${acc.total_balance:.2f}, Available: ${acc.available_balance:.2f}"
    )

    # Edge Case: Reserve over available balance
    acc.reserve(1000.0)
except (ValueError, TypeError) as e:
    print(f"Reservation Error: {e}")
```
