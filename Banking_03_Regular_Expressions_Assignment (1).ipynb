### Q1. Find a account code for banking case 1; use sample reference `ACC-1001` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
import re
from typing import Any, Dict, List, Optional
```

```python
def validate_account_code_v1(input_data: Any) -> Dict[str, Any]:
    """
    Validates account codes in the format 'ACC-' followed by exactly 4 digits.
    Example: ACC-1001
    """
    if not isinstance(input_data, str):
        return {"is_valid": False, "error": "Invalid type: Input must be a string.", "extracted": None}

    # Pattern: Matches literal 'ACC-', capturing 4 digits in a named group
    pattern = r"^ACC-(?P<account_number>\d{4})$"
    match = re.fullmatch(pattern, input_data)

    if match:
        return {"is_valid": True, "error": None, "extracted": match.groupdict()}
    return {"is_valid": False, "error": "Pattern mismatch: Must match ACC-XXXX.", "extracted": None}

# Test Cases
print("Q1 Tests:")
print(validate_account_code_v1("ACC-1001"))       # Valid
print(validate_account_code_v1("ACC-12"))         # Invalid (too short)
print(validate_account_code_v1("ABC-1001"))       # Invalid (wrong prefix)
print(validate_account_code_v1(""))               # Edge case: Empty input
print(validate_account_code_v1(1234))
```

```python
def extract_amounts_v1(text: Any) -> List[str]:
    """
    Extracts all currency amounts containing digits, commas, and decimals.
    Example: Finds '1,250.50' from a text log.
    """
    if not isinstance(text, str) or not text:
        return []

    # Matches sequences of digits separated by commas, ending with an optional decimal fraction
    pattern = r"\b\d{1,3}(?:,\d{3})*(?:\.\d{2})?\b"
    return re.findall(pattern, text)

print("\nQ3 Tests:")
print(extract_amounts_v1("Deposited 1,250.50 and withdrew 50.00 today.")) # ['1,250.50', '50.00']
```
