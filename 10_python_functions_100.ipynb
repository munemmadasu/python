Assignment 35: Fibonacci Function Write a function named generate_fibonacci().

Requirements Accept the number of terms. Return the complete Fibonacci series as a list. Handle zero and negative term counts. Do not use recursion in this assignment.

```python

def generate_fibonacci(n):
  if n<=0:
    return []
  if n==1:
    return [0]
  fib = [0,1]
  for i in range(2,n):
    fib.append(fib[-1]+fib[-2])
  return fib
generate_fibonacci(7)
```

Assignment 75: Filter Dates by Year Given dates in DD-MM-YYYY format:

Requirements Ask the user for a year. Use filter() to return dates belonging to that year. Handle invalid date strings safely. Use datetime.strptime() for validation.

```python

from datetime import datetime

dates = [
    "9-07-2004",
    "1-06-1994",
    "31-1-2024",
    "10-05-2023",
    "invalid-date",
    "3-09-2024",
]
year = input("Enter year: ")


def is_valid_and_matches(date_str):

    parts = date_str.split("-")
    if (
        len(parts) == 3
        and all(p.isdigit() for p in parts)
        and len(parts[2]) == 4
    ):
        d, m, y = map(int, parts)
        if 1 <= m <= 12 and 1 <= d <= 31:

            return datetime.strptime(date_str, "%d-%m-%Y").year == int(year)
    return False


result = list(filter(is_valid_and_matches, dates))
print(result)
```

```python

def categorize_experience(emp_list):
  def get_band(emp):
    updated_emp=emp.copy()
    exp=emp["experience"]
    if exp >=8:
      band="senior"
    elif exp>=4:
      band="mid-level"
    else:
      band="junior"
    updated_emp["experience_band"]=band
    return updated_emp
  return list(map(get_band,emp_list))
categorized_employees=categorize_experience(employees)
for emp in categorized_employees:
    print(
        f"Name: {emp['name']:<8} | Experience: {emp['experience']} yrs | Band: {emp['experience_band']}"
    )
```

```python


from functools import reduce
def department_salary(emp_list):
  departments=set(emp["department"] for emp in emp_list)
  dept_salaries={}
  for dept in departments:
    dept_emps=filter(lambda emp:emp["department"]== dept,emp_list)
    total_dept_salary=reduce(lambda x,y:x +y["salary"],dept_emps,0)
    dept_salaries[dept]=total_dept_salary
  return dept_salaries
result=department_salary(employees)
for dept,total in result.items():
  print(f"Department: {dept:<10} | Total Salary: ₹{total:,}")
```
