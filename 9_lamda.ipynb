```python
'''1. Using filter() and map():

1. Extract all even numbers.
2. Calculate the square of every even number.
3. Return the squared values as a list.

'''

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

even = filter(lambda x: x % 2 == 0, numbers)
result = list(map(lambda x: x ** 2, even))

print(result)
```

```python
'''2. Using filter() and reduce():

1. Extract all odd numbers.
2. Calculate their total sum.

'''

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

odd = filter(lambda x: x % 2 != 0, numbers)
total = reduce(lambda a, b: a + b, odd)

print(total)
```

```python
'''3. Using filter() and map():

1. Extract numbers greater than 50.
2. Multiply every selected number by 2.

'''

numbers = [10, 25, 40, 55, 70, 85, 100]

greater = filter(lambda x: x > 50, numbers)
result = list(map(lambda x: x * 2, greater))

print(result)
```

```python
'''4. Using filter(), map(), and reduce():

1. Extract positive numbers.
2. Calculate the square of every positive number.
3. Calculate the total sum of the squared values.

'''

numbers = [-15, 20, -35, 40, 50, -10, 60]

positive = filter(lambda x: x > 0, numbers)
squares = list(map(lambda x: x ** 2, positive))
total = reduce(lambda a, b: a + b, squares)

print(squares)
print(total)
```

```python
'''5. Using filter() and map():

1. Extract words containing more than five characters.
2. Convert the selected words to uppercase.

'''

words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]

result = list(map(lambda x: x.upper(),filter(lambda x: len(x) > 5, words)))

print(result)
```

```python
'''7. Using filter() and reduce():

1. Extract numbers divisible by both 3 and 5.
2. Calculate their total sum.

'''

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]

nums = filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers)
total = reduce(lambda a, b: a + b, nums)

print(total)
```

```python
'''8. Using filter() and reduce():

1. Extract palindrome words.
2. Join them into one string separated by " | ".

'''

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]

palindrome = filter(lambda x: x == x[::-1], words)
result = reduce(lambda a, b: a + " | " + b, palindrome)

print(result)
```

```python
'''9. Using filter(), map(), and reduce():

1. Extract temperatures greater than 30°C.
2. Convert the selected values into Fahrenheit.
3. Calculate the average Fahrenheit temperature.

'''

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

temp = list(map(lambda c: (c * 9/5) + 32,filter(lambda c: c > 30, temperatures)))

total = reduce(lambda a, b: a + b, temp)
average = total / len(temp)

print(temp)
print(average)
```

```python
'''12. Using filter() and reduce():

1. Extract employees from the IT department.
2. Calculate the total salary of IT employees.

'''

employees = [
    {"id": 101, "name": "Rahul", "department": "IT", "salary": 45000},
    {"id": 102, "name": "Anil", "department": "HR", "salary": 65000},
    {"id": 103, "name": "Sneha", "department": "IT", "salary": 72000},
    {"id": 104, "name": "Kiran", "department": "IT", "salary": 82000},
    {"id": 105, "name": "Meena", "department": "Sales", "salary": 55000}
]

it = filter(lambda e: e["department"] == "IT", employees)
total = reduce(lambda a, b: a + b["salary"], it, 0)

print(total)
```

```python
'''13. Using filter(), map(), and reduce():

1. Extract active employees.
2. Increase every active employee's salary by 15%.
3. Calculate the total updated payroll.

'''

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]

updated = list(map(lambda e: e["salary"] * 1.15, filter(lambda e: e["active"], employees)))

payroll = reduce(lambda a, b: a + b, updated)

print(updated)
print(payroll)
```

```python
'''14. Using filter() and reduce():

1. Extract students who passed.
2. Find the topper among the passed students.

'''

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

passed = filter(lambda s: s["passed"], students)

topper = reduce(lambda a, b: a if a["marks"] > b["marks"] else b, passed)

print(topper)
```

```python
'''15. Using filter(), map(), and reduce():

1. Extract students scoring more than 70.
2. Extract only their marks.
3. Calculate the average marks.

'''
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

marks = list(map(lambda s: s["marks"], filter(lambda s: s["marks"] > 70, students)))

total = reduce(lambda a, b: a + b, marks)
average = total / len(marks)

print(average)
```

```python
'''16. Using filter(), map(), and reduce():

1. Extract products currently in stock.
2. Calculate each product's inventory value using price × stock.
3. Calculate the total inventory value.

'''
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

values = list(map(lambda p: p["price"] * p["stock"], filter(lambda p: p["stock"] > 0, products)))

total = reduce(lambda a, b: a + b, values)

print(values)
print(total)
```

```python
'''19. Using filter(), map(), and reduce():

1. Extract transactions whose amount is greater than 5000.
2. Add 18% GST to every selected transaction.
3. Calculate the grand total after GST.

'''

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

gst = list(map(lambda t: t["amount"] * 1.18, filter(lambda t: t["amount"] > 5000, transactions)))

total = reduce(lambda a, b: a + b, gst)

print(gst)
print(total)
```

```python
'''20. Using filter(), map(), and reduce():

1. Extract active salespeople from the South region.
2. Calculate a 5% incentive for each selected salesperson.
3. Create a list containing salesperson name, sales amount, and incentive.
4. Calculate the total incentive payable.

'''

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

selected = filter(
    lambda s: s["region"] == "South" and s["active"],
    sales_records
)

result = list(map(
    lambda s: {
        "salesperson": s["salesperson"],
        "sales": s["sales"],
        "incentive": s["sales"] * 0.05
    },
    selected
))

total = reduce(lambda a, b: a + b["incentive"], result, 0)

print(result)
print(total)
```
