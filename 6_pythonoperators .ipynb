```python
#Calculate the total marks using += inside a loop.
marks=[90,80,70,60]
total_marks=0
for mark in marks:
  total_marks+=mark
  print(total_marks)
```

```python
#Compare two numbers and print the larger number.
a=10
b=20
if a>b:
  print(a)
else:
  print(b)
```

```python
#Check whether three numbers are equal.
a=50
b=50
c=50
if a==b==c:
  print("equal")
else:
  print("not equal")
```

```python
#Find the largest among three numbers using comparison operators.
a=20
b=100
c=50
if a>=b and a>=c:
  print(a)
elif b>=a and b>=c:
    print(b)
else:
    print(c)
```

```python
#Check whether a student is eligible to vote based on age.
age=18
if age>=18:
  print("eligible")
else:
    print("not eligible")
```

```python
#Logical Operators
```

```python
#Check if a person is eligible for a loan (age > 21 AND salary > 30000).
age=30
salary=30000
if age>=21 and salary>=30000:
  print("eligible")
else:
  print("not eligible")
```

```python
#Check if a number lies between 10 and 100 using logical operators.
number=50
if number>=10 and number<=100:
  print("lies")
else:
    print("not lies")
```

```python
#Validate username and password using logical operator
username="abc"
password=123
if username=="abc" and password==123:
  print("valid username and password")
else:
  print("invalid username and password")
```

```python
#Determine whether a year is a leap year using logical conditions.
year=2026
if year%4==0 and year%100!=0 or year%400==0:
  print("leap year")
else:
    print("not leap year")
```

```python
#Check whether a character exists in a string.
a=["data", "science"]
print("data"in (a))
```

```python
#Check whether a number exists in a list.
x=[1,23,56,2,8]
if 100 in x:
  print('exist')
else:
  print('not exist')
```

```python
#Count how many vowels in a string belong to a predefined vowel list using in.
x=("data")
y=["a","e","i","o","u"]
count=0
for i in x:
  if i in y:
    count+=1
    print(count)
```
