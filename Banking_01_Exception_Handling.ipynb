```python
class BankingError(Exception): """Base exception for all banking issues."""
class InsufficientFundsError(BankingError): """Raised when account balance is low."""
class InvalidTransactionError(BankingError): """Raised when a business validation rule fails."""
logging.basicConfig(level=logging.ERROR, format='%(levelname)s: %(message)s')
```

```python
#  Create InsufficientFundsError
def process_withdrawal(balance: float, amount: float) -> float:
    if amount > balance:
        raise InsufficientFundsError(f"[ACC-1031] Withdrawal of {amount} failed. Available balance: {balance}")
    return balance - amount
try:
    print(f"Normal Balance Check: Final Balance ${process_withdrawal(500.0, 100.0)}")
    process_withdrawal(100.0, 600.0)
except InsufficientFundsError as e:
    print(f"Funds Exception: {e}")

```

```python
def batch_process_transfers(transfers: list) -> dict:
    results = {"success": [], "failed": []}
    for tx in transfers:
        try:
            if tx.get("amount", 0) <= 0:
                raise InvalidTransactionError("Amount must be positive.")
            results["success"].append(tx["id"])
        except InvalidTransactionError as e:
            results["failed"].append({"id": tx.get("id"), "error": str(e)})
    return results
sample_batch = [{"id": "TXN_01", "amount": 150.0}, {"id": "TXN_02", "amount": -50.0}]
print(f"Batch Processing Logs: {batch_process_transfers(sample_batch)}")
```

```python

def execute_gateway_payment(payment_func, max_retries=3):
    for attempt in range(max_retries):
        try:
            return payment_func()
        except ConnectionError as e:
            if attempt == max_retries - 1:
                print(f"Gateway down after {max_retries} attempts.")
                raise e
            print(f"Attempt {attempt + 1} failed... Retrying connection.")
            time.sleep(0.01)
def broken_mock_gateway(): raise ConnectionError("Timeout hitting core settlement networks.")
try:
    execute_gateway_payment(broken_mock_gateway)
except ConnectionError:
    pass
```
