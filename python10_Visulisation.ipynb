```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, color='blue', marker='o', linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

Data Insights   
Upward Trend: The plot displays a strict, monotonically increasing trend. As the X value goes from 1 to 5, the Y value rises consistently from 10 to 40.Variable Slope / Growth Rate:X = 1 to 2: A steep rise (+10 units).X = 2 to 4: A gentler, steady climb (+5 units per step).X = 4 to 5: A final steep surge (+10 units).
Visual & Stylistic Elements
Data Markers: Blue circular points (marker='o') clearly highlight the exact measurements at each integer step along the X-axis (1, 2, 3, 4, 5).Line Style: A solid blue line (color='blue', linewidth=2) connects the data points, emphasizing continuity over time or sequential steps.Axes & Grid:The X-axis spans from 1.0 to 5.0, and the Y-axis spans from 10 to 40.Background grid lines (plt.grid(True)) align with tick marks, making it quick to estimate exact coordinates visually.

```python
plt.figure(figsize=(8, 5))
plt.plot(x, y, linestyle='--', color='red', label='Sales')
plt.legend()
plt.tight_layout()
plt.show()
```

Data Insights
Overall Trend: Represents steady upward movement (labeled as "Sales") from 10 to 40 across 5 time steps.

Rate of Increase:

Starts strong with a steep surge between step 1 and step 2 (+10).

Slows down into a moderate, uniform incline between steps 2 and 4 (+5 per step).

Accelerates sharply again between steps 4 and 5 (+10).

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
flights = sns.load_dataset('flights')

sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```


Visual & Technical Breakdown
Seaborn vs. Matplotlib: Seaborn calculates statistical aggregates and error bounds out of the box, avoiding the need to manually compute yearly averages or group the dataset.

Trade-off: While this plot clearly highlights the multi-year macro growth, grouping by year smooths over the fine-grained month-to-month fluctuations (e.g., individual summer peaks and winter dips).

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Bill Amount by Day')
plt.show()
```

Weekend vs. Weekday Bill Sizes:

Higher Spending on Weekends: Saturday and Sunday show higher median bill amounts (18–19) compared to Thursday and Friday (14–16).

Wider Variance on Weekends: The interquartile ranges (the height of the boxes) and whisker lengths are larger for Saturday and Sunday, showing that weekend diners spend across a much broader price range.

Outliers & High-Spending Parties:

Thursday, Saturday, and Sunday all feature individual outlier points above the upper whiskers.

Saturday has the highest individual outlier (~$50+), representing large groups or high-ticket orders.

Friday Dynamics:

Friday shows a relatively tight interquartile range with a lower median, suggesting consistent, lower-cost dining patterns (e.g., quick lunches or smaller evening groups).

```python
import plotly.express as px

df = px.data.iris()

fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

 Clear Species Separation (Setosa):

Iris setosa (blue dots) forms an isolated, linearly separable cluster on the top-left portion of the plot.
 It stands out with distinctly wider sepals (mostly 3.0 to 4.5 cm) paired with shorter sepal lengths (mostly 4.5 to 5.5 cm).

Overlapping Clusters (Versicolor vs. Virginica):

Iris versicolor (red dots) and Iris virginica (green dots) overlap in their sepal width ranges (typically 2.0 to 3.5 cm).

However, Virginica generally trends higher along the Y-axis (longer sepal length, mostly 6.0 to 7.9 cm) compared to Versicolor (mostly 5.0 to 7.0 cm).

Weak Overall Correlation Across All Data:

If you ignore species categories, sepal width and sepal length show very little overall linear correlation.

Within individual species (especially setosa), a positive trend emerges—demonstrating Simpson's Paradox, where a trend present within sub-groups disappears or reverses when aggregated.

```python
import plotly.express as px

df = px.data.gapminder().query("country == 'India'")

fig = px.line(df, x='year', y='lifeExp', title='Life Expectancy in India')
fig.update_traces(line=dict(color='green', width=3))
fig.show()
```

Key Data Insights

Steady Upward Trend: Life expectancy in India demonstrates a continuous, linear increase over time with no significant dips during the recorded periods.

Overall Growth: Between the early 1950s and mid-2000s, life expectancy rose from around 37–40 years to over 64 years—an increase of nearly 25+ years.

Consistent Progress: The slope shows steady, sustained improvements in public health, nutrition, healthcare infrastructure, and infant mortality reduction across successive decades.
Chart Specifications

X-Axis (year): Represents the years sampled in 5-year intervals (ranging from 1952 to 2007).

Y-Axis (lifeExp): Displays average life expectancy in years (spanning roughly 35 to 65 years).

Styling: Rendered as a solid green line with a line width of 3 using Plotly Express.
